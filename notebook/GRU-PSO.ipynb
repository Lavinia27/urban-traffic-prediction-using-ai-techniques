```python
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import matplotlib.pyplot as plt
from pandas.plotting import autocorrelation_plot
import pickle
import random, warnings, os
import datetime
import time 


warnings.filterwarnings('ignore')
np.random.seed(42)
random.seed(42)
torch.manual_seed(42)
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Rulăm pe: {device}")

print(torch.cuda.get_device_name(0))
```

```python
CSV_PATH = "PEMS-BAY.csv"
# Construirea secvențelor de intrare folosind o fereastră temporală
def build_dataset(data, lookback=12):
    X, y = [], []
    for i in range(len(data)-lookback):
        X.append(data[i:i+lookback])
        y.append(data[i+lookback])
    return np.array(X), np.array(y)


print("Încarc date...")
df = pd.read_csv(CSV_PATH).dropna()

data = df.iloc[:,1:].values.astype(np.float32)
print("Date shape:", data.shape)


LOOKBACK = 12

X, y = build_dataset(data, LOOKBACK)

print(f"Input shape (X): {X.shape}")
print(f"Output shape (y): {y.shape}")

# Împărțirea datelor în seturi de antrenare, validare și testare
n_samples = len(X)
train_end = int(0.7*n_samples)
val_end   = int(0.85*n_samples)

X_train, X_val, X_test = X[:train_end], X[train_end:val_end], X[val_end:]
y_train, y_val, y_test = y[:train_end], y[train_end:val_end], y[val_end:]

print("Train/Val/Test:", X_train.shape[0], X_val.shape[0], X_test.shape[0])
```

```python
# Normalizarea datelor în intervalul [0,1]
scaler_X = MinMaxScaler()
X_train_s = scaler_X.fit_transform(X_train.reshape(-1, X_train.shape[2])).reshape(X_train.shape)
X_val_s = scaler_X.transform(X_val.reshape(-1, X_val.shape[2])).reshape(X_val.shape)
X_test_s = scaler_X.transform(X_test.reshape(-1, X_test.shape[2])).reshape(X_test.shape)

scaler_y = MinMaxScaler()
y_train_s = scaler_y.fit_transform(y_train)
y_val_s   = scaler_y.transform(y_val)
y_test_s  = scaler_y.transform(y_test)

# Conversia datelor în tensori PyTorch
X_train_t = torch.tensor(X_train_s, dtype=torch.float32)
y_train_t = torch.tensor(y_train_s, dtype=torch.float32)
X_val_t   = torch.tensor(X_val_s, dtype=torch.float32)
y_val_t   = torch.tensor(y_val_s, dtype=torch.float32)
X_test_t  = torch.tensor(X_test_s, dtype=torch.float32)
y_test_t  = torch.tensor(y_test_s, dtype=torch.float32)
```

```python
# Definirea arhitecturii GRU
class GRUModel(nn.Module):
    def __init__(self, input_size, hidden_units, output_dim, dropout):
        super().__init__()
        self.gru = nn.GRU(
            input_size=input_size,
            hidden_size=hidden_units,
            num_layers=2, 
            batch_first=True,
            dropout=dropout if dropout > 0 else 0
        )
        self.fc = nn.Linear(hidden_units, output_dim)

    def forward(self, x):  # Se utilizează ultima ieșire temporală pentru realizarea predicției
        out, _ = self.gru(x)
        out = self.fc(out[:, -1, :])
        return out
```

```python
# Antrenarea modelului pentru evaluarea unei configurații candidate
def train_model_gru(model, lr, batch_size, epochs=10):
    criterion = nn.MSELoss() 
    optimizer = optim.Adam(model.parameters(), lr=lr)
    dataset = TensorDataset(X_train_t, y_train_t)
    loader = DataLoader(dataset, batch_size=int(batch_size), shuffle=True)
    model.to(device)

    for epoch in range(epochs):
        model.train()
        for xb, yb in loader:
            xb, yb = xb.to(device), yb.to(device)
            optimizer.zero_grad()
            pred = model(xb)
            loss = criterion(pred, yb)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)  # Limitarea normei gradientului pentru stabilizarea antrenării
            optimizer.step()
            
  # Evaluarea configurației pe setul de validare
    model.eval()
    with torch.no_grad():
        # Mutăm setul de validare pe device
        X_val_device = X_val_t.to(device)
        y_val_device = y_val_t.to(device)
        val_pred = model(X_val_device)
        val_loss = criterion(val_pred, y_val_device)
        
    return val_loss.item()
```

```python
def fitness_function_gru(params):
# Conversia parametrilor optimizați în hiperparametrii modelului
    hidden_units = int(params[0])
    lr = 10**params[1]
    dropout = np.clip(10**params[2], 0.0, 0.4)
    batch_size = [16, 32, 64][np.argmin(np.abs(np.array([16, 32, 64]) - params[3]))]
    
    model = GRUModel(X_train_s.shape[2], hidden_units, y_train_s.shape[1], dropout)
    mse_val = train_model_gru(model, lr, batch_size, epochs=15)

    model.eval()
    with torch.no_grad():
        pred_val = model(X_val_t.to(device)).cpu().numpy()
    
    std_real = np.std(y_val_s)
    std_pred = np.std(pred_val)
    variance_penalty = abs(std_real - std_pred)
    
    return mse_val + (0.5 * variance_penalty)
```

```python
class PSO_Hyper:

    def __init__(self, func, dim=4, population=10, max_iter=5):
        self.func = func
        self.dim = dim
        self.population = population
        self.max_iter = max_iter
        self.lb = np.array([32, -4.5, -2.0, 16]) 
        self.ub = np.array([256, -2.0, -0.5, 64]) 
        self.v_max = (self.ub - self.lb) * 0.2
        self.positions = np.random.uniform(self.lb,self.ub,(population,dim))
        self.velocities = np.zeros((population,dim))
        self.pbest = self.positions.copy()
        self.pbest_val = np.array([func(p) for p in self.positions])
        best_idx = np.argmin(self.pbest_val)
        self.gbest = self.pbest[best_idx].copy()
        self.gbest_val = self.pbest_val[best_idx]
        self.history = []

    def optimize(self):
        c1, c2 = 1.5, 1.5
        for it in range(self.max_iter):
            w = 0.9 - (0.5 * (it / self.max_iter)) 
            for i in range(self.population):
                r1,r2 = np.random.rand(self.dim), np.random.rand(self.dim)
                self.velocities[i] = (
                    w*self.velocities[i]
                    + c1*r1*(self.pbest[i]-self.positions[i])
                    + c2*r2*(self.gbest-self.positions[i])
                )
                self.velocities[i] = np.clip(self.velocities[i], -self.v_max, self.v_max)
                self.positions[i] += self.velocities[i]
                self.positions[i] = np.clip(self.positions[i],self.lb,self.ub)
                fit = self.func(self.positions[i])
                
                if fit < self.pbest_val[i]:
                    self.pbest[i] = self.positions[i].copy()
                    self.pbest_val[i] = fit

                if fit < self.gbest_val:
                    self.gbest = self.positions[i].copy()
                    self.gbest_val = fit
            self.history.append(self.gbest_val)

            print(f"Iter {it+1}/{self.max_iter} | Best MSE: {self.gbest_val:.6f}")
        return self.gbest, self.gbest_val, self.history

```

```python
print("\n Încep optimizarea PSO pentru modelul GRU...")
start = time.time()
pso = PSO_Hyper(fitness_function_gru, population=10, max_iter=5) 
best_h, best_fit, history = pso.optimize()
print(f" Done PSO in {(time.time()-start)/60:.2f} minute")
```

```python
hidden_units_opt = int(best_h[0])
lr_opt = 10**best_h[1] 
dropout_opt = np.clip(10**best_h[2], 0.0, 0.4)
batch_options = [16, 32, 64]
batch_opt = batch_options[np.argmin(np.abs(np.array(batch_options) - best_h[3]))]

print("  REZULTATE OPTIMIZARE PSO (MODEL GRU)")
print(f"  Unități Hidden   : {hidden_units_opt}")
print(f"  Learning Rate    : {lr_opt:.6f}")
print(f"  Dropout          : {dropout_opt:.4f}")
print(f"  Batch Size       : {batch_opt}")
print(f"  Best Fitness Scor: {best_fit:.6f} (MSE + Var_Penalty)")
```

```python
# Parametrii utilizați pentru antrenarea finală
hidden_units_opt = int(best_h[0])
lr_opt = 10**best_h[1]
dropout_opt = np.clip(10**best_h[2], 0.0, 0.4)
batch_options = [16, 32, 64]
batch_opt = batch_options[np.argmin(np.abs(np.array(batch_options) - best_h[3]))]

print(f"\n Antrenament Final GRU: hidden={hidden_units_opt}, lr={lr_opt:.6f}, dropout={dropout_opt:.6f}, batch={batch_opt}")
start_train = time.time()

# Inițializarea modelului și a optimizatorului
model = GRUModel(
    input_size=X_train_s.shape[2],
    hidden_units=hidden_units_opt,
    output_dim=y_train_s.shape[1],
    dropout=dropout_opt
).to(device)

criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=lr_opt)
dataset = TensorDataset(X_train_t, y_train_t)
loader = DataLoader(dataset, batch_size=batch_opt, shuffle=True)

num_epochs = 50 

for epoch in range(num_epochs):
    model.train()
    epoch_loss = 0
    for xb, yb in loader:
        xb, yb = xb.to(device), yb.to(device) 
        optimizer.zero_grad()
        pred = model(xb)
        loss = criterion(pred, yb)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0) # Limitarea gradientului pentru stabilizarea antrenării
        optimizer.step()
        epoch_loss += loss.item()

    if (epoch + 1) % 10 == 0:
        model.eval()
        with torch.no_grad():
            val_pred = model(X_val_t.to(device))
            val_loss = criterion(val_pred, y_val_t.to(device))
        print(f"Epoch: {epoch+1}/{num_epochs} | Train Loss: {epoch_loss/len(loader):.6f} | Val Loss: {val_loss.item():.6f}")

print(f" Antrenament finalizat in {(time.time()-start_train)/60:.2f} minute")
```

```python
# Evaluarea modelului pe setul de testare
model.eval()
with torch.no_grad():
    X_test_device = X_test_t.to(device)
    pred_test = model(X_test_device).cpu().numpy()

# Revenirea la valorile originale
pred_test_inv = scaler_y.inverse_transform(pred_test)
y_test_inv = scaler_y.inverse_transform(y_test_s)

# Calculul metricilor de evaluare
mse_real  = mean_squared_error(y_test_inv, pred_test_inv)
rmse_real = np.sqrt(mse_real)
mae_real  = mean_absolute_error(y_test_inv, pred_test_inv)
r2_real   = r2_score(y_test_inv, pred_test_inv)

print("\n REZULTATE FINALE GRU-PSO:")
print(f"MSE  : {mse_real:.4f}")
print(f"RMSE : {rmse_real:.4f}")
print(f"MAE  : {mae_real:.4f} ")
print(f"R2   : {r2_real:.4f} ")
```
